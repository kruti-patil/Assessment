```python
'''
Restaurant Ordering System
Design a Python class RestaurantSystem to simulate a basic restaurant order
management process. The system should support:
Functional Requirements:
➔ Display Menu:
◆ Show a list of food items with prices and item codes.
➔ Take Order:
◆ Allow users to select multiple items by item code and quantity.
◆ Calculate total price and generate a unique order ID.
➔ View Order:
◆ Look up details of an order using the order ID.
➔ Cancel Order:
◆ Cancel an existing order using the order ID.
➔ Exit the system
'''
```

```python
class RestaurantSystem:
  def __init__(self):
    self.menu = {
            101: {"name": "Pizza", "price": 250},
            102: {"name": "Burger", "price": 120},
            103: {"name": "Pasta", "price": 200},
            104: {"name": "Fries", "price": 80},
            105: {"name": "Coke", "price": 50},
        }
    self.orders = {}
    self.next_id = 1
  def display_menu(self):
    print("\n--- MENU ---")
    for code, item in self.menu.items():
        print(f"{code}: {item['name']} - ₹{item['price']}")
    print("------------")

  def take_order(self):
    order_items = []
    total = 0
    while True:
      try:
          code = int(input('Enter item code (0 to stop):'))
          if code == 0:
            break
          if code in self.menu:
              qty = int(input(f"Enter quantity for {self.menu[code]['name']}: "))
              if qty > 0 :
                 unit_price = self.menu[code]['price']
                 subtotal = unit_price * qty
                 order_items.append((self.menu[code]['name'], qty, unit_price, subtotal))
                 total += subtotal
                 print(f"Added \033[1m{qty} x {self.menu[code]['name']} (₹{unit_price} each) = ₹{subtotal}\033[0m")
              else:
                print("\033[1mInvalid quantity.\033[0m")
          else:
            print("Invalid item code. Please try again.")
            self.display_menu()
      except ValueError:
        print("Invalid input. Please enter a number.")
        self.display_menu()

    if order_items:
      order_id = self.next_id
      self.next_id += 1
      self.orders[order_id] = {"items": order_items, "total": total, "status": "Active"}
      # print("\n--- BILL ---")
      # for n, q, price, sub in order_items:
      #   print(f"{n} x{q} (₹{price}) = ₹{sub}")
      # print(f"\033[1mTOTAL = ₹{total}\033[0m")
      print(f"\n Order placed! Order ID: \033[1m{order_id}\033[0m")
    else:
      print(" No items were ordered.")

  def view_order(self):
      oid_str = input("Enter Order ID (0 to view all, or comma-separated IDs): ").strip()
      if oid_str == "0":
        active_orders = {oid: o for oid, o in self.orders.items() if o["status"] != "Cancelled"}
        if not active_orders:
            print("No orders found.")
            return
        print("\n--- All Orders ---")
        for order_id, o in active_orders.items():
            print(f"\nOrder ID: {order_id} | Status: {o['status']}")
            for n, q, price,sub in o["items"]:
              print(f"{n} x{q} (₹{price}) = \033[1m₹{sub}\033[0m")
            print(f" \033[1m Total = ₹{o['total']}\033[0m")
        print("-------------------")
        return
      try:
        ids = [int(x.strip()) for x in oid_str.split(",") if x.strip().isdigit()]
        if not ids:
          print("\033[1mInvalid input. Please enter numbers only.\033[0m")
          return
        for oid in ids:
          if oid in self.orders:
              o = self.orders[oid]
              if o["status"] == 'Cancelled':
                print(f"\033[1m {oid} order has been cancelled.\033[0m")
                continue
              print(f"\n\033[1m--- Order {oid} ---\033[0m")
              for n, q, price,sub in o["items"]:
                print(f"{n} x{q} (₹{price}) = ₹{sub}")
              print(f"\033[1m Total = ₹{o['total']} | Status = {o['status']}\033[0m")
          else:
              print(f"Order {oid} not found.")
      except ValueError:
          print("\033[1mInvalid Order ID format.\033[0m")

  def cancel_order(self):
    try:
        oid = int(input("Enter Order ID to cancel: "))
        if oid in self.orders and self.orders[oid]["status"] == "Active":
            self.orders[oid]["status"] = "Cancelled"
            print(f"\033[1mOrder {oid} has been cancelled.\033[0m")
        else:
            print("\033[1mOrder not found or already cancelled.\033[0m")
    except ValueError:
        print("\033[1mInvalid Order ID.\033[0m")

  def main_menu(self):
      while True:
          print("\n1. Display Menu")
          print("2. Take Order")
          print("3. View Order")
          print("4. Cancel Order")
          print("5. Exit")
          choice = input("Enter your choice: ")

          if choice == "1":
              self.display_menu()
          elif choice == "2":
              self.take_order()
          elif choice == "3":
              self.view_order()
          elif choice == "4":
              self.cancel_order()
          elif choice == "5":
              print("\033[1m Thank you!\033[0m")
              break
          else:
              print("Invalid choice. Try again.")


```
